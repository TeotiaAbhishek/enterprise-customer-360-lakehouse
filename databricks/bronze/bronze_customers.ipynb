```python
raw_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/workspace/default/data/crm_customers_20260616_154420.csv")
)

display(raw_df.limit(10))
```

```python
raw_df.printSchema()
```

```python
(
    raw_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("bronze_customers")
)
```

```python
spark.sql("""
SELECT *
FROM bronze_customers
LIMIT 10
""")
```
