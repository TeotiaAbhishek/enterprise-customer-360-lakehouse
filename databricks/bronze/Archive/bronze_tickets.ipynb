```python
raw_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/workspace/default/data/support_tickets_20260616_162156.csv")
)

display(raw_df.limit(10))
```

```python
raw_df.printSchema()
```

```python
(
    raw_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("bronze_tickets")
)
```
